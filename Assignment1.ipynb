```python

students_db = {}

def add_student(roll_no, name, branch, marks_tuple):
    """1. Add a new student record."""
    if roll_no not in students_db:

        students_db[roll_no] = [name, branch, marks_tuple]
        print(f"[+] Added student: {name} (Roll No: {roll_no})")
    else:
        print(f"[-] Roll No {roll_no} already exists.")

def delete_student(roll_no):
    """2. Delete an existing student record."""
    if roll_no in students_db:
        removed_student = students_db.pop(roll_no)
        print(f"[-] Deleted student: {removed_student[0]} (Roll No: {roll_no})")
    else:
        print(f"[!] Cannot delete. Roll No {roll_no} not found.")

def update_student(roll_no, new_name=None, new_branch=None, new_marks=None):
    """3. Update the details of a student."""
    if roll_no in students_db:
        if new_name:
            students_db[roll_no][0] = new_name
        if new_branch:
            students_db[roll_no][1] = new_branch
        if new_marks:
            students_db[roll_no][2] = new_marks
        print(f"[*] Updated details for Roll No: {roll_no}")
    else:
        print(f"[!] Cannot update. Roll No {roll_no} not found.")

def display_records():
    """4. Display the final student records."""
    print("\n" + "="*40)
    print("FINAL STUDENT RECORDS")
    print("="*40)
    if not students_db:
        print("No student records available.")
    else:
        for roll, details in students_db.items():
            name, branch, marks = details
            print(f"Roll No : {roll}")
            print(f"Name    : {name}")
            print(f"Branch  : {branch}")
            print(f"Marks   : {marks}")
            print("-" * 25)




add_student(2404, "Pauravee Phuke", "Computer Science", (85, 90, 88))
add_student(102, "Shreya", "Electrical Eng", (75, 82, 80))
add_student(103, "Rohan Gupta", "Mechanical Eng", (91, 85, 89))


delete_student(103)


update_student(2404, new_branch="Artificial Intelligence", new_marks=(95, 92, 98))


display_records()
```
